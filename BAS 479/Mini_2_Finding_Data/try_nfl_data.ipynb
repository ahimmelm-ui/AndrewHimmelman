## Maybe we try some NFL Data?

```python
import nflreadpy as nfl
import pandas as pd
```

```python
# nflreadpy initially returns a Polars DataFrame
nfl_data = nfl.load_pbp([2025])

# Convert to pandas if that is what your class uses
nfl_data = nfl_data.to_pandas()

nfl_data.shape
```

```python
# Lets build a data frame that I can use for a model on play call predictions

final_data = nfl_data[[
    "game_date",
    "game_id",
    "play_type",
    "down",
    "ydstogo",
    "yardline_100",
    "qtr",
    "game_seconds_remaining",
    "score_differential",
    "posteam_timeouts_remaining",
    "defteam_timeouts_remaining",
    "goal_to_go",
    "posteam",
    "defteam",
    "season_type",
]].copy()

final_data = final_data[final_data["play_type"].isin(["run", "pass"])]
```

```python
final_data = final_data.dropna()
final_data.shape
```

```python
final_data.to_csv("nfl_play_data.csv", index=False)
```
